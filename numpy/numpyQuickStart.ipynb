```python
#创建0维数组
# np.array(5)它的第一个参数接收"任何类似数组的输入"——包括列表、元组、嵌套结构，也包括单个标量。传 5 时，NumPy 把 5 包装成一个"只有一个元素、没有维度"的数组：
# 类构造函数
arr_zero = np.array(5)
#dtype = data_type
print(arr_zero.dtype)
print(arr_zero.ndim)
print(arr_zero.shape)
print(arr_zero.size)
print(arr_zero)
print(arr_zero.ctypes)
```

```python
# 常用创建函数，创建不同数据矩阵变量
arange = np.arange(10)
# 创建 shape为 2行3列 value 全为0的 二维数组,返回值  NDArray[float64]: 代表
zeros = np.zeros((3,1))
# 创建 shape 为5行4列 value 全为1的 二维数组
ones = np.ones((5,4))
print(zeros.ndim)
print(zeros.shape)
print(zeros.size)

print(ones.ndim)
print(ones.shape)
print(ones.size)
```

```python
#预测概率
predictions = np.array([0.9,0.3,0.8,0.2,0.7])
#标识
labels = np.array([1,0,1,0,1])

# 将连续概率预测转为二分类标签：预测值 > 0.5 判定为正类(1)，否则为负类(0)
# .astype(int) 将布尔值(True/False)转换为整数(1/0)，方便与真实标签比较
# 二分类标签
binary_pred = (predictions > 0.5).astype(int)

# 逐元素比较预测标签与真实标签是否一致，得到布尔数组
# np.mean 计算 True 所占比例，即为预测正确的样本比例（准确率）
accuracy = np.mean(binary_pred == labels)
#打印准确率，:.2% 格式化为保留两位小数的百分比形式，如 87.50%
print(f'准确率：{accuracy:.2%}')
#统计计算
#打印预测值数组的标准差
print(f'均值：{predictions.mean():.3f}')
#打印预测值数组的标准差
print(f'标准值：{predictions.std():.3f}')
#预测值数组的最大值
print(f'最大值：{predictions.max()}')
```

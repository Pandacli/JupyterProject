```python

import pandas as pd
import numpy as np
print(pd.__version__)
```

```python
#创建一个pandas 的一维数组
series = pd.Series(['abc','23',3,4,5,6])
print(series)
```

```python
#创建一个 DataFrame，包含 一个 datetime index 且 使用date_range() 作为labeled column


#随机生成6个日期，格式如 xxxx-xx-xx
dates = pd.date_range(start='2019-01-01', periods=6)
dates
print(type(dates))
# data参数 = 随机生成(6,4)的值


# 服从标准正态分布（也叫高斯分布）的随机数。标准正态分布：均值（μ）= 0，标准差（σ）= 1。参数 (6,4)：
# 指定生成的随机数数组的形状为 6 行 4 列（二维数组）。
random_data = np.random.randn(6,4)
# index参数 = dates 日期
# columns 列 = 把ABCD字符串转换为列表
dataFrame=pd.DataFrame(random_data, index=dates, columns=list('ABCD'))
print(dataFrame)
```

```python
#创建dictionary key-value数据，key为列数据， value 为行数据
dictionary = {
    "A": 1.0,
    #创建Timestamp object
    "B": pd.Timestamp("20130102"),
    #创建Series object
    "C": pd.Series(1,index=list(range(4)),dtype='float32'),
    "D": np.array([3]*4,dtype='int32'),
    #创建Categorical object
    "E": pd.Categorical(['test','train','test','train']),
}
df2 = pd.DataFrame(dictionary)
print(df2)
print(df2.dtypes)
```

```python
#列的顺序按列名排序(从D 到A )，行顺序不变，降序排序
dataFrame.sort_index(axis=1, ascending=False)
#列的顺序按列名排序(从A 到D )，行顺序不变，降序排序
dataFrame.sort_index(axis=1, ascending=True)
#行的顺序按行名排序(从01-06 到 01-01 )，列顺序不变，倒序
dataFrame.sort_index(axis=0, ascending=False)
```

```python
dataFrame.sort_values(by='A')
#设置特定某一列进行排序
dataFrame.sort_values(by='A', ascending=False)
```

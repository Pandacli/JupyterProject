```python
#元祖
data= ('赵信','盖伦','莉莉娅','孙悟空','刘德华','梁朝伟')
print(len(data))
print(data[0])
print(data[0:2])
#获取 索引位置1 开始往后的元素
print(data[1:])
#获取 索引位置 倒数-1 开始 往前的元素
print(data[:-1])
#步长
#从索引 1 开始，到索引 4 结束（不包含 4），每 2 步取一个元素。
print(data[1:4:2])
#从索引0开始，到最后，每2步取一个元素
print(data[0::2])
print(data[1::2])
#步长为-1，从右往左取，从索引4开始，索引1结束。
print(data[4:1:-1])
```

```python
#1.通过{}填充，创建
set1 ={1,2,3}
print(set1,type(set1))

#2 通过构造方法 set(iterable) 遍历
set2 = set('hogwarts')
print(set2,type(set2))

set3 = set([1,2,3,4])
print(set3,type(set3))

set4 = set()
print(set4,type(set4))

#3.通过集合推导式,in 判断元素是否在集合中存在，如果i%2 ==0 符合条件的数填充到集合
set5= { i for i in range(5) if i%2==0}
print(set5,type(set5))


#判断元素是否在集合中存在 in ， not in
#遍历当前集合
set6 ={1,2,3,4,5,67,100,231}
a= 25
print(a in set6)
print(a not in set6)
newSet = {index for index in range(len(set6)) if set6.pop() %a ==0}
print(newSet)
```

```python
#定义一个函数，将a+b,python 函数无需强制定义数据类型
def add(a,b):
    return a+b

#假设实参 是整数 4,5 。 则return 4+5 ，并打印其type ，查看返回值类型是？
result = add(4,5)
print(type(result))
#假设实参是 字符串，则拼接字符串，返回值类型应该是 str
result2 = add('hello',' world')
print(type(result2))
#假设实参是 浮点数，则相加后 得到的是浮点数
result3 = add(5.1,10)
print(result3)
print(type(result3))

#极端尝试 使用Ture+False  ,返回值 = True & False ，逻辑返回1 ?
#解释：Python 中布尔类型（bool）是整数类型（int）的子类，False 本质是 0 的 “布尔别名”，True 是 1 的 “布尔别名”

result4 = add(False,False)
print(result4)
print(type(result4))


result5 = add(False,True)
print(result5)
print(type(result5))

#验证这个继承关系和数值等价性
print
```

```python
def stop_view(name,something,age):
    print(f'停止观看：{name}的AV,她最喜欢的姿势:{something},她的年龄是{age}')
    
stop_view('新有菜','狗仔式',30)
#错误示例：
stop_view(30,'新有菜','传教士')
```

```python
#限制传参方式
#代表 / 前面的name 参数只能用关位置参数
#* 后面的参数 只能用关键字参数
def greet(name,/,gender,*,age,height):
    print(f'{name},{gender},{age},{height}')

#正确。
greet('tommy','男',age=30,height=180)QQQQQQ
#正确.
greet('genis',gender='female',age=30,height=180)
#错误
#greet(name='tommy',gender='male',age=30,height=180)
#错误
greet('tommy',gender='male',30,height=180)
```

```python
#可变位置参数
def index_args(arg1,arg2, arg3=19,*args):
    print(f"this is{arg1} ,{arg2}index args function")
    print(*args)
index_args(1,2,30,12,102,123210,231,321)
```

```python
#定义一个 可变位置参数 *args ， 参数
def my_sum(*args):
    return sum(args)

print(my_sum(1, 2, 3))       # 6
print(my_sum(10, 20, 30, 40)) # 100
```

```python
def example(**kwargs):
    print(type(kwargs))   # <class 'dict'>
    print(kwargs)         # 例如 {'name': 'Alice', 'age': 25}

example(name="Alice", age=25)
#将关键字参数 name = 'Alice',age = 25 传入到 **kwargs 形参的 字典中
```

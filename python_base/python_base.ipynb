```python
#列表的推导式
#类似java 的List<Double> passed = scores.stream().filter(s -> s > 0.8).collect(Collectors.toList());
# passed 是个 列表 ，遍历获取 s>0.8 的数值，重新放入到新的 列表去。
passed = [s for s in scores if s > 0.8]


#字典推导式 ，用法类比 Java 8+ Stream API 对 Map.entrySet() 进行操作，并使用 Collectors.toMap()
# results.items() 返回键值对元组
# 输出是一个新的字典，遍历所有键值对，其中键 (k) 保持不变，值 (v) 被替换为布尔值（True 如果 v > 0.9，否则 False）。
result_map = {key: value > 0.9 for key, value in results.items()}
```

```python
#大模型评估，输入 预测、标签、阈值
def evaluate_model(predictions, labels, threshold=0.5):
   correct = sum(1 for p, l in zip(predictions, labels)
            if (p >= threshold) == l)
   result = correct / len(labels)
   print(result)
   return result


```

```python
import pandas as pd


# 读取数据集 2分
data = pd.read_csv('sensor_data.csv')
```

```python
# 1. 传感器数据统计
# 对传感器类型进行分组，并计算每个组的数据的数量和平均值 3分
# 分组后获取GroupBy 对象，对['Value']列 使用agg数据聚合计算 mean 平均值，count 计数
sensor_stats = data.groupby('SensorType')['Value'].agg(['count', 'mean'])
# 输出结果
print("传感器数据数量和平均值:")
print(sensor_stats)
```

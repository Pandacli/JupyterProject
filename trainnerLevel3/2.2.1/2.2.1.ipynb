```python
import pandas as pd
# 加载数据
data = pd.read_csv('finance数据集.csv')
# 显示前五行的数据
data.head(5)
print(data.head(5))
```

```python
#第二小题
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import pickle
# 选择特征: 自变量和因变量
#
# 变量x = 移除无用无用索引列 && 机器学习的目标列，并将删除后的新 DataFrame 赋值给变量 X（通常用于机器学习的特征数据）。个人在过去两年内是否出现过严重的拖欠（1 表示有严重拖欠，0 表示没有），索引号
x = data.drop(['SeriousDlqin2yrs', 'Unnamed: 0'], axis=1)
#变量Y = 个人在过去两年内是否出现过严重的拖欠（1 表示有严重拖欠，0 表示没有）
y = data['SeriousDlqin2yrs']

# 分割训练集和测试集（测试集20%）
#X_train, X_test, y_train, y_test = __________(__________, random_state=42)
# 分割
X_train, X_test, y_train, y_test =train_test_split (x,y,test_size=0.2 ,random_state=42)
##
# 训练Logistic回归模型（最大迭代次数为1000次）
model = LogisticRegression(max_iter=10000)
#训练 Logistic 回归模型
model.fit(X_train, y_train)
#y_pred = model.predict(X_test)

# 保存模型
with open('2.2.1_model.pkl', 'wb') as file:
    #pickle.__________
    pickle.dump(model, file)

# 使用测试集X_test 来预测y的结果，并创建DataFrame并保存结果
y_pred = model.predict(X_test)
pd.DataFrame(y_pred, columns=['预测结果']).to_csv('2.2.1_results.txt', index=False)
```

```python
#第三小题
from sklearn.metrics import classification_report
# 生成测试报告
report = classification_report(y_test, y_pred, zero_division=1)
with open('2.2.1_report.txt', 'w') as file:
    file.write(report)
# 分析测试结果
# 测试集 与 预测集 结果一致性
result = (y_test==y_pred)
#获取 结果一致性的平均值
accuracy = result.mean()
#保存两位小数点
print(f"模型准确率: {accuracy:.2f}")
```

```python
#第五小题
# 运用工具分析算法中错误案例产生的原因并进行纠正，重新得到模型训练结果，以文件名2.2.1_results_xg.txt保存到考生文件夹。

from imblearn.over_sampling import SMOTE
#重采样
smote = SMOTE(random_state=42)
x_resampled,y_resampled = smote.fit_resample(X_train,y_train)
#重新训练
model.fit(x_resampled, y_resampled)
#重新预测
y_pred_resampled = model.predict(X_test)
#将预测结果保存到2.2.1_results.xg.txt
pd.DataFrame(y_pred_resampled,columns=['重采样预测结果']).to_csv('2.2.1_results.xg.txt', index=False)
report_resampled = classification_report(y_test, y_pred_resampled, zero_division=1)
with open("2.2.1_results_xg.txt",'w')as file:
    file.write(report_resampled)

#分析重采样后的 测试结果
accuracy_resampled = (y_test==y_pred_resampled).mean()
#保留两位小数
print(f"模型准确率: {accuracy_resampled:.2f}")
```

```python
import pandas as pd
import numpy as np
from bs4.builder import FAST

# 读取数据集 1分
#返回对象是 DataFrame 对象[具有横竖列的数据结构] or TextFileReader
data = pd.read_csv('patient_data.csv')
```

```python
# 1. 统计 住院天数 超过7天的患者 数量 及其 占比
# 创建新列'RiskLevel'，根据住院天数判断风险等级 3分
#np.where(condition,x,y)  x 为 condition为True时返回的值，y为condition为False时返回的值 ，    Return elements chosen from `x` or `y` depending on `condition`.
daysInHospital= data['DaysInHospital']
#高风险表示；低风险标识
riskLevel= np.where(daysInHospital>7,'高风险患者','低风险患者')
data['RiskLevel']=np.where(daysInHospital>7,'高风险患者','低风险患者')
# 统计不同风险等级的患者数量 2分
risk_counts = data.groupby('RiskLevel').size()
#risk_counts = data['RiskLevel'].value_counts()

#计算高风险患者占比 1分
risk_high_risk_ratio = risk_counts['高风险患者'] / len(data['RiskLevel'] )

#计算低风险患者占比 1分
risk_low_risk_ratio = risk_counts['低风险患者']/ len(data['RiskLevel'])


print("高风险患者数量:", risk_counts['高风险患者'])
print("低风险患者数量:", risk_counts['低风险患者'])
print("高风险患者占比:", risk_high_risk_ratio)
print("低风险患者占比:", risk_low_risk_ratio)
```

```python
# 2. 统计不同BMI区间中高风险患者的比例和统计不同BMI区间中的患者数
# 定义BMI区间和标签
bmi_bins = [0, 18.5, 24, 28, np.inf]
bmi_labels = ['偏瘦', '正常', '超重', '肥胖']
# 根据BMI值划分指定区间 4分
# pd.cut(x,bins,labels)  将x 根据bins划分区间，返回区间标签bmi_labels
data['BMIRange'] = pd.cut(data['BMI'], bmi_bins, bmi_labels,)
#data['BMIRange'] = _____________(_____________, _____________, _____________, right=False)  # 使用左闭右开区间
# 计算每个BMI区间 高风险患者的比例 2分
# 1）按照每个BMI区间 分组,分组结果 再统计 apply 高风险者的比例
rate = lambda x : (x=='高风险患者').mean()
# 先分组，再获取具体列名'RiskLevel'，再获取列名对应的值
bmi_risk_rate= data.groupby('BMIRange',observed=False)['RiskLevel'].apply(rate,include_groups=False).sort_index(ascending=False)
# 统计每个BMI区间的患者数量 1分
bmi_patient_count = data['BMIRange'].value_counts().sort_index(ascending=True)

# 输出结果
print("BMI区间中高风险患者的比例和患者数:")
print(bmi_risk_rate) 
print(bmi_patient_count)
```

```python
# 3. 统计不同年龄区间中高风险患者的比例和统计不同年龄区间中的患者数
# 定义年龄区间和标签
age_bins = [0, 26, 36, 46, 56, 66, np.inf]
age_labels = ['≤25岁', '26-35岁', '36-45岁', '46-55岁', '56-65岁', '＞65岁']
# 根据年龄值划分指定区间 4分
data['AgeRange'] = pd.cut(data['Age'], age_bins, age_labels)
#print(data)
#data['AgeRange'] = _____________(_____________, _____________, _____________, right=False)  # 使用左闭右开区间
# 计算每个年龄区间中高风险患者的比例 2分
# select age_risk_rate from data where RiskLevel='高风险患者' group by AgeRange
age_risk_rate = data.groupby('AgeRange',observed=False)['RiskLevel'].apply(lambda x: (x == '高风险患者').mean())
#age_risk_rate = _____________(_____________)['RiskLevel'].apply(lambda x: (x == '高风险患者').mean())
# 统计每个年龄区间的患者数量 1分, 约等于 select AgeRange from data group by AgeRange
age_patient_count = data['AgeRange'].value_counts(ascending=True)
#age_patient_count = data_____________

# 输出结果
print("年龄区间中高风险患者的比例和患者数:")
print(age_risk_rate)
print(age_patient_count)
```
